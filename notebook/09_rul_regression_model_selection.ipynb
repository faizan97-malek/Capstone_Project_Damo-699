## Why it was replaced

## What notebook 11 does instead

Notebook 11 builds a **physics-based synthetic RUL target** that incorporates all four
sensor variables:

- **Tool wear** (primary degradation signal)
- **Torque** (mechanical stress — high torque accelerates wear)
- **Temperature difference** (thermal stress — process temp minus air temp)
- **RPM inverse** (higher RPM = lower torque but faster rotation wear)

The degradation rate formula maps directly to the AI4I failure modes (TWF, HDF, PWF, OSF),
producing a TTF that responds realistically to all sensor inputs. A sensitivity analysis
in notebook 11 confirms that increasing torque lowers TTF, high temperature lowers TTF,
and a low-stress combination gives higher TTF — as expected physically.

The trained model is saved to `models/rul_regressor.joblib` and loaded by
`src/inference.py` for live TTF estimates on the dashboard.
